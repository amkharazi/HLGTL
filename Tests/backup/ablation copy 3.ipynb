# Synthetic Data

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


radius = 2

def label_points_inside_sphere(samples, radius=10):
    # Calculate the distance of each point from the origin
    distances = np.linalg.norm(samples, axis=1)
    
    # Label points based on whether they are inside the sphere
    labels = (distances < radius).astype(int)
    
    return labels

# Generate the 3D tensor data
n_samples = 1000
size = (n_samples, 32, 32, 32)
samples = np.random.normal(loc=0, scale=np.sqrt(3), size=size)

# Reshape the data to 2D for plotting
samples_2d = samples.reshape(n_samples, -1)

# Label points inside the sphere
labels = label_points_inside_sphere(samples_2d)

# Create a 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the data points in 3D with colors based on labels
for idx, sample in enumerate(samples_2d):
    color = 'b' if labels[idx] == 1 else 'r'
    ax.scatter(sample[0], sample[1], sample[2], c=color, marker='o')

# Create a sphere boundary
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(0, np.pi, 100)
x = radius * np.outer(np.cos(u), np.sin(v))
y = radius * np.outer(np.sin(u), np.sin(v))
z = radius * np.outer(np.ones(np.size(u)), np.cos(v))
ax.plot_surface(x, y, z, color='g', alpha=0.2)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Plot with Sphere Boundary')

plt.show()
```

```python
class simple_linear(nn.Module):
    def __init__(self):
        super(simple_linear, self).__init__()
        self.fc = nn.Linear(in_features= 1, out_features=1)
    def forward(self, x):
        x = x.reshape(-1,1)
        return self.fc(x)
    
model = simple_linear().to(device)

# Train for 100 epochs
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)


num_epochs = 100
for epoch in range(num_epochs):
    # Forward pass
    y_pred = model(x)
    loss = criterion(y_pred, y.unsqueeze(dim = 1))
    
    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    if (epoch+1) % 1 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')


y_pred = model(torch.tensor(x_true, dtype=torch.float32, device=device)).detach().cpu().numpy()

plt.figure(figsize=(8, 6))
plt.scatter(x_true, y_true)
plt.plot(x_true, y_pred)
plt.show()
```
