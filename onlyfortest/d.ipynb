```python
from skimage import data
import matplotlib.pyplot as plt
import torch
import sys
```

```python
sys.path.append('..') 
from Utils.Reshape import reshape
```

```python
astronaut = data.astronaut()
astronaut_tensor = torch.tensor(astronaut)

plt.imshow(astronaut)
plt.axis('off') 
plt.show()
```

```python
print("Shape of the image tensor:", astronaut_tensor.shape)
```

```python
astronaut_tensor = astronaut_tensor.unsqueeze(0)
print("Modified shape of the image tensor:", astronaut_tensor.shape)
```

```python
result = reshape(split=[2,2,1], map_type=1)
```

# map type1 

```python
reshaper = reshape(split=[2, 2, 1], map_type=1) 

output_tensor = reshaper(astronaut_tensor)
print(output_tensor.shape)
```

```python
# Display the specified subset
for i in range(2):
    for j in range(2):
        subset = output_tensor[0, i, j, 0, :, :, :]
        plt.subplot(2, 2, i * 2 + j + 1)
        plt.imshow(subset.numpy())
        plt.axis('off')

plt.show()
```
