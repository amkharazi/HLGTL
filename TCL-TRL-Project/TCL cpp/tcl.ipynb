```python
from setuptools import setup, Extension
from torch.utils.cpp_extension import CppExtension, BuildExtension

setup(
    name='tensor_contraction_cpp',
    ext_modules=[
        CppExtension('tensor_contraction_cpp', ['tensor_contraction.cpp']),
    ],
    cmdclass={'build_ext': BuildExtension}
)
```

```python
import torch
from torch import nn
from torch.utils.cpp_extension import load

# Load the custom C++ extension
tensor_contraction_cpp = load(name="tensor_contraction_cpp", sources=["tensor_contraction.cpp"])

class TensorContraction(nn.Module):
    def __init__(self, R1, R2, R3, C, H, W):
        super(TensorContraction, self).__init__()
        self.A = nn.Parameter(torch.randn(R1, C))
        self.B = nn.Parameter(torch.randn(R2, H))
        self.C = nn.Parameter(torch.randn(R3, W))
    
    def forward(self, T):
        return tensor_contraction_cpp.tensor_contraction(T, self.A, self.B, self.C)

# Example usage
# B, C, H, W = batch size, channels, height, width
# R1, R2, R3 = sizes of resulting matrices
model = TensorContraction(R1=4, R2=5, R3=6, C=3, H=32, W=32)
input_tensor = torch.randn(2, 3, 32, 32)  # Example input
result = model(input_tensor)
print(result.shape)  # Should print torch.Size([2, R1, R2, R3])
```
